## Task 1 — Treemap: fossil fuel dependency by country

**What to build:** A treemap showing **fossil fuel TWh only**, broken down by Region → Country → Source (Coal / Oil / Natural Gas).

**Requirements:**
- Filter to fossil sources only before plotting
- Use `path=['Region', 'Country', 'Source']` for the hierarchy
- Colour encodes the fossil source type (Coal / Oil / Natural Gas) with a CVD-safe palette
- Show TWh values in labels — no percentages
- Grey out parent nodes (Region and Country level)
- Insight title naming which region or country is most fossil-dependent

> 💡 `df.loc[df['Source_Type'] == 'Fossil']`


```python
# Task 1: fossil fuel dependency by country
df_fossil = df.loc[df['Source_Type'] == 'Fossil'].copy()
fossil_by_country = df_fossil.groupby('Country')['TWh'].sum().sort_values(ascending=False)
top_country = fossil_by_country.index[0]
top_region = df_fossil.groupby('Region')['TWh'].sum().idxmax()
fossil_colors = {'Coal': '#2E75B6', 'Oil': '#E07B39', 'Natural Gas': '#009E73'}
fig1 = px.treemap(df_fossil, path=['Region', 'Country', 'Source'], values='TWh', color='Source',
                  color_discrete_map=fossil_colors, labels={'TWh': 'Energy (TWh)'},
                  title=f'{top_region} has the largest fossil total; {top_country} leads among countries',
                  template='plotly_white', height=720)
tr = fig1.data[0]
tr.marker.colors = [('#AAAAAA' if str(node_id).count('/') < 2 else fossil_colors.get(label, '#AAAAAA'))
                   for node_id, label in zip(tr.ids, tr.labels)]
fig1.update_traces(texttemplate='%{label}<br>%{value:,.0f} TWh',
                   hovertemplate='<b>%{label}</b><br>%{value:,.0f} TWh<extra></extra>', root_color='#AAAAAA')
fig1.update_layout(margin=dict(t=75,l=10,r=10,b=10))
fig1.show()

```

## Task 2 — Treemap vs bar: low-carbon energy by country

**What to build:** Build **both** a treemap and a horizontal bar chart showing total low-carbon TWh (Nuclear + Hydro) per country. Then answer the question in a markdown cell below.

**Requirements:**
- Filter to `Source_Type == 'Low-carbon'` and aggregate TWh by country
- Treemap: single-level `path=['All', 'Country']` with a dummy root node labelled `'Low-carbon'`
- Bar chart: sorted by TWh, horizontal orientation, CVD-safe colour
- Both charts show TWh values, not percentages
- Insight title on the bar chart naming the leading country

```python
# Task 2: low-carbon energy by country
df_low = df.loc[df['Source_Type'] == 'Low-carbon']
low_by_country = df_low.groupby('Country', as_index=False)['TWh'].sum().sort_values('TWh', ascending=False)
leader = low_by_country.iloc[0]['Country']
low_by_country['All'] = 'Low-carbon'
fig2a = px.treemap(low_by_country, path=['All', 'Country'], values='TWh',
                    color_discrete_sequence=['#2E75B6'], labels={'TWh': 'Energy (TWh)'},
                    title='Low-carbon energy by country', template='plotly_white', height=600)
fig2a.update_traces(texttemplate='%{label}<br>%{value:,.0f} TWh',
                    hovertemplate='<b>%{label}</b><br>%{value:,.0f} TWh<extra></extra>', root_color='#AAAAAA')
fig2a.show()
fig2b = px.bar(low_by_country.sort_values('TWh'), x='TWh', y='Country', orientation='h',
               color_discrete_sequence=['#2E75B6'], text='TWh',
               labels={'TWh': 'Low-carbon energy (TWh)'},
               title=f'{leader} leads low-carbon energy production', template='plotly_white', height=600)
fig2b.update_traces(texttemplate='%{x:,.0f} TWh', textposition='outside', cliponaxis=False)
fig2b.update_layout(showlegend=False, yaxis_title='')
fig2b.show()
```

### Task 2 — reflection

**Which chart tells the ranking story more clearly, and why?**

The horizontal bar chart tells the ranking story more clearly: all countries share a baseline, and ordered bar lengths make even close values easier to compare. The treemap is useful for seeing each country's relative contribution to the total, but area is harder to compare precisely.

## Task 3 — 100% stacked bar: which region relies most on coal?

**What to build:** A horizontal 100% stacked bar chart showing, for each **Region**, how its fossil energy splits between Coal, Oil and Natural Gas. This is the **same data as your Task 1 treemap**, visualised a second way.

**Requirements:**
- Reuse `df_fossil` from Task 1; aggregate TWh by `Region` × `Source` and convert to **shares of each region's fossil total**
- Horizontal bars, regions **sorted by coal share** (highest at the top)
- Coal is the segment on the baseline (first in `category_orders`) and the **only coloured segment** — keep it the same colour as in Task 1; Oil and Natural Gas in two greys
- Percentage labels on the coal segments only
- Insight title naming the most coal-dependent region and its coal share

> 💡 `df.groupby('Region')['TWh'].transform('sum')` gives each row its region total

```python
# Task 3: regional fossil mix as 100% stacked horizontal bars
regional_mix = df_fossil.groupby(['Region', 'Source'], as_index=False)['TWh'].sum()
regional_mix['Share'] = regional_mix['TWh'] / regional_mix.groupby('Region')['TWh'].transform('sum')
coal_share = regional_mix.loc[regional_mix['Source'] == 'Coal'].set_index('Region')['Share'].sort_values(ascending=False)
region_order = coal_share.index.tolist()
most_coal_region, coal_pct = coal_share.index[0], coal_share.iloc[0]
regional_mix['Label'] = np.where(regional_mix['Source'] == 'Coal', regional_mix['Share'].map('{:.0%}'.format), '')
regional_colors = {'Coal': '#2E75B6', 'Oil': '#AAAAAA', 'Natural Gas': '#DDDDDD'}
fig3 = px.bar(regional_mix, x='Share', y='Region', color='Source', text='Label', orientation='h',
             category_orders={'Region': region_order, 'Source': ['Coal', 'Oil', 'Natural Gas']},
             color_discrete_map=regional_colors, labels={'Share': 'Share of regional fossil energy'},
             title=f'{most_coal_region} relies most on coal ({coal_pct:.0%} of fossil energy)',
             template='plotly_white', height=450)
fig3.update_traces(textposition='inside', insidetextanchor='start', textfont=dict(color='white'),
                  marker_line_color='white', marker_line_width=1)
fig3.update_layout(xaxis=dict(tickformat='.0%', range=[0,1]), yaxis_title='', legend_title='Fossil source')
fig3.show()
```

### Task 3 — reflection

**Your Task 1 treemap and this stacked bar show the same fossil data. Name one question each chart answers that the other cannot.**

The treemap answers where the largest absolute fossil fuel volumes are and how they break down from region to country to source. The 100% stacked bar answers which region has the highest coal share of its fossil mix, making regional composition easy to compare.
